3.2 What is Data Cleaning?
Data Cleaning (or Data Preprocessing) is the process of identifying and correcting inaccurate,
incomplete, duplicate, or inconsistent data to improve its quality before analysis.
Raw data often contains:
• Missing values
• Duplicate records
• Incorrect spellings
• Extra spaces
• Different letter cases
• Invalid data
• Wrong formats
Without cleaning, analytical results and machine learning models may become inaccurate or
unreliable.

3.3 Importance of Data Cleaning
Data cleaning is important because it:
• Improves data quality.
• Increases machine learning model accuracy.
• Eliminates redundant information.
• Reduces errors in analysis.
• Enhances decision-making.
• Ensures data consistency.
• Saves time during analysis.

3.4 Common Data Quality Issues
A. Missing Values
Missing values occur when some data is unavailable.
Example:
Name Age
John 25
Alice NaN
David 30
Reasons:
• Human error
• Sensor failure
• Data corruption
• Incomplete forms

Methods to Handle Missing Values
Method 1: Remove Missing Records

```python
df.dropna()
```

Useful when only a few records contain missing values.

Method 2: Replace with Mean

```python
df["Age"].fillna(df["Age"].mean())
```

Suitable for numerical data without significant outliers.

Method 3: Replace with Median

```python
df["Age"].fillna(df["Age"].median())
```

Suitable when outliers are present.

Method 4: Replace with Mode

```python
df["City"].fillna(df["City"].mode()[0])
```

Suitable for categorical data.

B. Duplicate Records
Duplicate records represent repeated observations.
Example:
ID Name
101 John
101 John
Duplicates can bias statistical analysis and machine learning.

Detect duplicates:

```python
df.duplicated()
```

Remove duplicates:

```python
df.drop_duplicates()
```

C. Inconsistent Data
Inconsistent data occurs due to different spellings, cases, or formatting.
Example:
Bangalore
bangalore
BANGALORE
Bengaluru
All represent the same city.

Cleaning methods:

```python
str.lower()
str.upper()
str.title()
str.strip()
```

 Read CSV dataset

```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv("employees.csv")

print("Original Dataset")
print(df)

print("\nDataset Information")
print(df.info())
print("\nMissing Values")
print(df.isnull().sum())
# Fill missing numerical values with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
# Fill missing city names
df["City"] = df["City"].fillna("Unknown")
# Remove duplicate rows
df = df.drop_duplicates()
# Remove extra spaces
df["City"] = df["City"].str.strip()
# Convert city names to title case
df["City"] = df["City"].str.title()
print("\nCleaned Dataset")
print(df)
# Save cleaned dataset
df.to_csv("cleaned_employees.csv", index=False)

print("\nCleaned dataset saved successfully.")
```
