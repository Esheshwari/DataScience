3. Theory
3.1 Introduction to Data Wrangling
Data Wrangling (also known as Data Munging) is the process of converting raw, unstructured,
or inconsistent data into a clean and organized format suitable for analysis, visualization, and
machine learning.
In real-world applications, raw datasets often contain:
• Missing values
• Duplicate records
• Incorrect formats
• Unnecessary columns
• Inconsistent values
• Unorganized structures
Data wrangling improves the quality and usability of the data by performing various
transformation operations.

3.2 Importance of Data Wrangling
Data wrangling is essential because it:
• Improves data quality.
• Makes datasets easier to analyze.
• Enhances machine learning model performance.
• Removes inconsistencies and redundancy.
• Supports accurate business decision-making.
• Simplifies visualization and reporting.

3.3 Data Transformation
Data transformation refers to modifying the structure, format, or values of data to meet
analytical requirements.
Common transformation operations include:
• Filtering
• Sorting
• Grouping
• Aggregation
• Reshaping
• Renaming columns
• Creating new features

3.4 Filtering Data
Filtering is used to select only those records that satisfy a specified condition.
Example
Suppose a company wants to identify employees earning more than ₹45,000.

3.5 Sorting Data
Sorting arranges records in ascending or descending order.
Ascending Order
df.sort_values("Salary")
Descending Order
df.sort_values("Salary", ascending=False)
Sorting helps identify:
• Highest salary
• Lowest sales
• Top-performing products
• Latest transactions

3.6 Grouping Data
Grouping combines records having similar values.
Example:
City Salary
Delhi 40000
Delhi 45000
Mumbai 50000
Grouping by City:
Delhi
Mumbai
Syntax

df.groupby("City")

3.7 Aggregation
Aggregation summarizes grouped data.
Common aggregation functions include:
Function Description
sum() Total
mean() Average
max() Highest value
min() Lowest value
count() Number of records
median() Middle value
std() Standard deviation
Example:
df.groupby("City")["Salary"].mean()
Output:
City Average Salary
Delhi 42500
Mumbai 50000

3.8 Reshaping Data
Reshaping changes the structure of a dataset.
Common techniques include:
Pivot Table
Converts rows into columns.
pd.pivot_table(df,
values="Salary",
index="Department",
columns="City",
aggfunc="mean")

Melt
Converts columns into rows.
pd.melt(df)

```python
import pandas as pd

# Read dataset
df = pd.read_csv("employees.csv")

print("Original Dataset")
print(df)
# ------------------------------
# Filtering
# ------------------------------
print("\nEmployees with Salary > 45000")
filtered = df[df["Salary"] > 45000]
print(filtered)

# ------------------------------
# Sorting
# ------------------------------

print("\nSorted by Salary (Descending)")
sorted_df = df.sort_values(by="Salary", ascending=False)
print(sorted_df)
# ------------------------------
# Grouping
# ------------------------------
print("\nGrouped by Department")
grouped = df.groupby("Department")
# ------------------------------
# Aggregation
# ------------------------------
print("\nAverage Salary by Department")
print(grouped["Salary"].mean())
print("\nMaximum Salary by Department")
print(grouped["Salary"].max())
print("\nEmployee Count by Department")
print(grouped["Salary"].count())
# ------------------------------
# Pivot Table
# ------------------------------
print("\nPivot Table")
pivot = pd.pivot_table(
df,
values="Salary",
index="Department",
columns="City",
aggfunc="mean"
)
print(pivot)
# ------------------------------
# Melt
# ------------------------------

print("\nMelted Dataset")
melted = pd.melt(
df,
id_vars=["ID", "Name"],
value_vars=["Department", "City"]
)
print(melted)
```
