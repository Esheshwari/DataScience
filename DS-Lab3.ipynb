**3. Theory**

3.1 Introduction to ETL

ETL stands for Extract, Transform, and Load. It is a data integration process used to collect
data from multiple sources, clean and transform it, and store it in a target system such as a
database or data warehouse.

ETL is widely used in:

• Business Intelligence (BI)
• Data Warehousing
• Data Migration
• Data Analytics
• Machine Learning
• Cloud Data Engineering

The ETL process ensures that raw data is converted into high-quality, structured information
suitable for analysis and decision-making.

3.2 Phases of ETL

A. Extract

The Extract phase retrieves data from one or more data sources.

Common data sources include:

• CSV files
• Excel spreadsheets
• Relational databases (MySQL, PostgreSQL)
• NoSQL databases (MongoDB)
• APIs
• Cloud storage
• Web scraping

Example:
df = pd.read_csv("employees.csv")
Here, data is extracted from a CSV file into a Pandas DataFrame.

B. Transform

The Transform phase cleans, processes, and modifies the extracted data.

Typical transformation tasks include:

• Handling missing values
• Removing duplicate records
• Correcting inconsistent entries
• Changing data types
• Filtering records
• Creating new columns
• Aggregating data
• Standardizing text

Example:
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
This replaces missing salary values with the average salary.

C. Load

The Load phase stores the transformed data into the destination.

The destination may be:

• CSV file
• SQL database
• Data warehouse
• Cloud database
• Big Data platform

Example:
df.to_csv("cleaned_data.csv", index=False)
or
df.to_sql("employees", connection)

**7. Python Program**

```python
import pandas as pd
import sqlite3
# -----------------------------
# Extract Phase
# -----------------------------
print("Reading Raw Data...")
df = pd.read_csv("employees.csv")
print("\nOriginal Dataset")
print(df)
# -----------------------------
# Transform Phase
# -----------------------------
print("\nTransforming Data...")
# Handle missing values
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Remove duplicate records
df = df.drop_duplicates()

# Remove leading/trailing spaces
df["City"] = df["City"].str.strip()

# Standardize city names
df["City"] = df["City"].str.title()
print("\nCleaned Dataset")
print(df)
# -----------------------------
# Load to CSV
# -----------------------------
df.to_csv("cleaned_employees.csv", index=False)
print("\nCleaned CSV file created.")
# -----------------------------
# Load to SQLite Database
# -----------------------------
connection = sqlite3.connect("employee.db")
df.to_sql(
"Employees",
connection,
if_exists="replace",
index=False
)
print("\nData Loaded into SQLite Database Successfully.")
# Verify Database
query = "SELECT * FROM Employees"
result = pd.read_sql(query, connection)
print("\nDatabase Records")
print(result)
connection.close()
```
