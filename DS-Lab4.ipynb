**3. Theory**

3.1 Introduction to Data Integration

Data Integration is the process of combining data from multiple heterogeneous sources into a
single, consistent, and meaningful dataset. It enables organizations to obtain a unified view of
information for reporting, analytics, and decision-making.

In today's organizations, data is generated from multiple departments and systems such as:

• Sales
• Human Resources
• Finance
• Customer Relationship Management (CRM)
• Enterprise Resource Planning (ERP)
• E-commerce platforms
• IoT devices
• Cloud applications

Since these systems store data in different formats and locations, integration is required before
meaningful analysis can be performed.

3.2 What is ETL?

ETL stands for:

• Extract – Collect data from different sources.
• Transform – Clean, validate, standardize, and integrate the data.
• Load – Store the processed data into a target system such as a database, data warehouse,
or CSV file.

3.5 Data Validation

consistent, and ready for analysis.
Validation includes:

• Identifying duplicate records.
• Verifying data types.
• Ensuring key relationships.
• Checking record counts.
• Validating merged data.
Example:
If an employee belongs to Department ID 101, that department should exist in the department
dataset.

**7. Python Program**

```python
import pandas as pd

# -----------------------------
# Extract Phase
# -----------------------------

# Employee dataset
employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105, 106, 106],
    "Name": ["John", "Alice", "Raj", "David", "Riya", "Sam", "Sam"],
    "DepartmentID": [1, 2, 2, 3, 1, 3, 3],
    "Salary": [35000, 42000, 50000, 55000, 45000, None, None]
})

# Department dataset
departments = pd.DataFrame({
    "DepartmentID": [1, 2, 3],
    "DepartmentName": ["HR", "IT", "Finance"],
    "Location": ["Bangalore", "Mumbai", "Delhi"]
})

print("Employee Dataset")
print(employees)

print("\nDepartment Dataset")
print(departments)

# -----------------------------
# Transform Phase
# -----------------------------

# Merge datasets
merged = pd.merge(
    employees,
    departments,
    on="DepartmentID",
    how="inner"
)

# Handle missing values
merged["Salary"] = merged["Salary"].fillna(
    merged["Salary"].mean()
)

# Remove duplicate records
merged = merged.drop_duplicates()

print("\nIntegrated Dataset")
print(merged)

# -----------------------------
# Validation
# -----------------------------

print("\nMissing Values")
print(merged.isnull().sum())

print("\nDuplicate Records")
print(merged.duplicated().sum())

print("\nData Types")
print(merged.dtypes)

print("\nNumber of Records")
print(len(merged))

# -----------------------------
# Load Phase
# -----------------------------

merged.to_csv(
    "Integrated_Employee_Data.csv",
    index=False
)

print("\nIntegrated Dataset Saved Successfully.")
```
